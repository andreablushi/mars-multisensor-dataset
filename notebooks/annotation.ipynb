# Annotation

This notebook checks the evaluation tiles by eye on the THEMIS mosaic. Each class
lists all its labelled tiles in the order the balanced draw takes them, past the
number drawn, so a rejected tile is replaced by the next one down. Each click
writes its verdict to `data/analysis/verdicts.json` straight away.

## Setup

```python
"""Import the labels, the verdicts so far, and what reviews them."""

from analysis.ground_truth import catalogue
from analysis.visualization.ground_truth import review, tables
from common.config import analysis_settings

labels = catalogue.read_labels()
ground_truth = analysis_settings().ground_truth
verdicts = catalogue.read_verdicts()
```

## Review

Accept or reject each tile for its class. A class opens on its first tile not yet
reviewed.

```python
"""Judge every labelled tile of a class, in draw order."""

review.plot(labels, ground_truth, verdicts)
```

## Second review

Only the accepted tiles, to reject any that slipped through. Run the cell again to
include tiles accepted since.

```python
"""Judge the accepted tiles again."""

review.plot(labels, ground_truth, verdicts, accepted_only=True)
```

## Where the review stands

Run the cell again after reviewing. "Drawn now" is the draw the current verdicts
give, and "Drawn, not reviewed" is what is left to check of it. Upload the verdicts
with the command in the README, then run the analysis again.

```python
"""Tabulate every class's verdicts and the draw they leave."""

tables.verdicts(labels, ground_truth, verdicts)
```
