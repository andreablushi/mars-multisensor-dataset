# What the filter would make of the dataset

This notebook runs the feature selection filter written in `configs/window_filter.yaml` over the features computed on disk and reads what it would leave.

## Setup

```python
"""Import the published stats and the selection, and the tables that read them."""

from analysis import dataset_list
from analysis.stats.artifacts import store
from analysis.stats.dataset import catalogue
from analysis.visualization.dataset.tables import (
    classes,
    index,
    landing,
    size,
)
from building.configs import overall
```

## The ODE dataset

How much there is to choose from, before the filter is asked of it. Every
number here is read off the measurement itself, so it is what the rest of the
notebook is judged against.

### The ODE dataset that was measured

Read straight off the coverage measurement, so it stands whatever the filter
goes on to keep.

```python
"""Read the catalogue index, which every table in this section is drawn from."""

measured = catalogue.read_catalogue()
index.measured(measured)
```

## Sweep the features

What the pipeline published for the filter as it is written now, read back
rather than swept again.

```python
"""Read what the stats pipeline published for every computed feature."""

read = store.read_stats_file()
```

### Global instrument coverage

One row per instrument, over the whole measured record and before the filter.

```python
"""Tabulate what each instrument holds of the measured features."""

index.instruments(measured, read)
```

## Filling a feature

A window is searched over one whole feature at a time, so the feature is the unit
everything below is measured on. This section focuses on what a feature is ideally
expected to cover.

### What each instrument lands on a feature

Whether an instrument brings a feature enough to be worth asking it for anything.

It's important to note that the observations offered come from before the filter, so they are a global overview, while the coverage reached is over the features that earned a window.

```python
"""Tabulate what each instrument lands on a feature and how far it reaches."""

landing.landed(read)
```

## Final dataset size

What the filter would hand over: the features it keeps of the features it
searched, how long their windows run, and how many observations each
instrument offered them.

### The dataset the filter would leave

```python
"""Tabulate the dataset the filter would leave behind."""

size.final(read)
```
