# What the filter would make of the dataset

This notebook runs the tile selection filter written in `configs/window_filter.yaml` over the tiles computed on disk and reads what it would leave.

## Setup

```python
"""Import the published selection and stats, and what reads them."""

from analysis.stats.artifacts import selection, store
from analysis.stats.dataset import catalogue
from analysis.visualization.dataset.plots import tiles
from analysis.visualization.dataset.tables import (
    index,
    landing,
    size,
)
```

## The ODE dataset

How much there is to choose from, before the filter is asked of it. Every
number here is read off the measurement itself, so it is what the rest of the
notebook is judged against.

### The ODE dataset that was measured

Read straight off the coverage measurement, so it stands whatever the filter
goes on to keep.

```python
"""Read the catalogue index, which every table in this section is drawn from."""

measured = catalogue.read_catalogue()
index.measured(measured)
```

## Sweep the tiles

What the pipeline published for the filter as it is written now, read back
rather than swept again.

```python
"""Read what the stats pipeline published for every computed tile."""

read = store.read_stats_file()
```

### Global instrument coverage

One row per instrument, over the whole measured record and before the filter.

```python
"""Tabulate what each instrument holds of the measured tiles."""

index.instruments(measured, read)
```

## Filling a tile

A window is searched over one whole tile at a time, so the tile is the unit
everything below is measured on. This section focuses on what a tile is ideally
expected to cover.

### What each instrument lands on a tile

Whether an instrument brings a tile enough to be worth asking it for anything.

It's important to note that the observations offered come from before the filter, so they are a global overview, while the coverage reached is over the tiles that earned a window.

```python
"""Tabulate what each instrument lands on a tile and how far it reaches."""

landing.landed(read)
```

## Final dataset size

What the filter would hand over: the tiles it keeps of the tiles it
searched, how long their windows run, and how many observations each
instrument offered them.

### The dataset the filter would leave

```python
"""Tabulate the dataset the filter would leave behind."""

size.final(read)
```

### The tiles kept on Mars

Every tile on the THEMIS mosaic, green where the filter kept it and red where it did not.

```python
"""Map every tile of Mars, the kept ones in green and the excluded ones in red."""

tiles.plot(selection.read_selection())
```
