# Coverage of a Mars tile

Pick a point, confirm, and see what every instrument has observed of it.

- **Per-observation coverage** shows what a single observation covered, as a share of the
  tile's bounding box, at the time it was taken, with the window the search chose
  shaded across it.
- **Cumulative coverage** shows how much of the tile each instrument has reached in
  total, as its observations accumulate.
- **Pixels per observation** shows how many pixels one look of an instrument lands on
  the tile, against the pixels the filter asks of it there.
- **What the tile holds** shows the window it earned, and what each instrument
  left inside it, in ground and in the pixels that ground is worth.

### What is measured

**Tile's bounding box:** Mars is split into equal-area tiles of about 32 km a
side, each a south/north/west/east box, so the box is what coverage is measured
against.

**The ground is flattened:** A degree is not a fixed amount of ground: near the poles it covers far less than at the equator. So the same patch measured in degrees would look bigger or smaller depending only on where it happens to sit. Everything is therefore laid onto a flat map drawn for that one place, of the kind that keeps every patch its true size wherever it falls. Areas can then just be added up, and a share means the same thing at any latitude.

**CTX and CRISM publish areas:** ODE gives each of these products a footprint
polygon, so the observation is taken as published: cut to the tile box in lon/lat,
resampled and projected. The resolution of the product is not considered, just the area it covers.

**MOLA excluded:** Since it's coverage is full planet, it is not useful to show it here.

**SHARAD publishes a bare line:** It is a sounder, so ODE gives the ground track it flew and no width at all. A width is needed before a line can become area, and rather than pick one number for the whole mission, every track derives its own through mission metadata.


### What is left out

**Only targeted sets with both a footprint and a time are downloaded:** An observation
with no acquisition time cannot be placed on a time axis at all, and a whole-planet
basemap reads full coverage on every tile by construction while saying nothing about
targeted observing. Neither kind is downloaded, computed, or plotted here.

**An observation is kept when it overlaps at all:** A footprint that only partly reaches
into the tile is cropped to it and counted for the part that reaches. Only an
observation with no overlap whatsoever is discarded.

### Setup

```python
"""Import the pipeline's own picker, plots, and tables."""

from analysis.visualization.common.picker import TilePicker
from analysis.visualization.tile.plots import (
    basemap,
    cumulative,
    footprints,
    observations,
    pixels,
)
from analysis.visualization.tile.tables import held
```

## Select a tile

Type a latitude and longitude, and the tile holding that point is loaded. Whether it
has local data is noted, as is whether the search kept it or left it without a window.

A window is one stretch of time no longer than the filter allows. The search slides it
over everything the instruments observed of the tile and asks, at each position, how
much ground the observations inside it reach. Of the windows that meet every share the
filter insists on it keeps the one worth the most, ground weighed against the days it
has to wait for it, and the tile earns no place when none of them meets the filter.


```python
"""Show the tile picker, which fills every area claimed below it."""

picker = TilePicker()
picker.choose()
```

### Load the confirmed tile's coverage

The mosaic is overlaid with the ground the tile covers, and beside it a note says
whether a window was found and what the filter asks of one.
The mosaic drawn under every tile is THEMIS daytime infrared, served as WMS by the
USGS.

```python
"""Report the tile and show the ground it covers."""

picker.show_panel(basemap.plot)
```

## The tile as a whole

Everything below is measured over the whole tile, which is what the search now runs
over and what one sample of a training dataset would hold.

### Per-observation coverage

One point per observation, at the time it was taken, each dropped to the axis by a thin
stem so a single observation stays visible where the points crowd together. The height is
the share of the tile's bounding box that one observation covered, so a tall point is a
wide swath and a low point is a narrow one.

The stretch the window is open over is shaded across it, except on the panel of an
instrument the filter asks of the whole record, which no window binds.

Zero-coverage points are plotted since they are still observations taken in consideration by ODE, however the intersection is minimal.

```python
"""Plot what each single observation covered, one panel per instrument."""

picker.show_panel(observations.plot)
```

### Cumulative coverage

It shows how much of the tile each instrument has reached in total, as its observations accumulate through time. 

```python
"""Plot how much of the tile each instrument reaches over time, and in total."""

picker.show_panel(cumulative.plot)
```

### Pixels per observation

How many pixels one look lands on the tile, together with a vertical line showing the pixel bound of the filter.


```python
"""Plot how many pixels one observation lands on the tile."""

picker.show_panel(pixels.plot)
```

### What the tile holds

The tile itself, the window it earned, and what each instrument left on it in ground
and in the pixels that ground is worth. 

```python
"""Summarise what the search left on the tile."""

picker.show_panel(held.plot)
```

### The footprints it keeps

The tile's own crop of the mosaic, with the ODE footprint of every observation it keeps
traced on it, one colour per instrument set.

```python
"""Map the tile with the footprint of every observation its window keeps."""

picker.show_panel(footprints.plot)
```
